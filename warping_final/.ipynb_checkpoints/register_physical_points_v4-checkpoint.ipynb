```python
# projection into sphere
# ring_0_proj (center) - 0
# ring_1_proj
# ring_2_proj
# ring_3_proj
# ring_4_proj - excl indces 7,8,9 (out of sphere)
# ring_5_proj - excl indces 7,8,9 (out of sphere) (seam)
# ring_6_proj - excl indces 7,8,9 (out of sphere)
# ring_7_proj - excl indces 7,8,9 (out of sphere)

# groundtruth image:
# ring_0_gt (center)
# ring_1_gt
# ring_2_gt
# ring_3_gt
# ring_4_gt - excl indces 7,8,9 (match above)
# ring_5_gt - excl indces 7,8,9 (match above)
# ring_6_gt - excl indces 7,8,9 (match above)
# ring_7_gt - excl indces 7,8,9 (match above)

# out of canvas points that should not be included in mapping
excl_indices = {4:(7,8,9), 5:(7,8,9), 6:(7,8,9), 7:(7,8,9)}
n_rings = 8
```

```python
# aggregate registered points in dictionary with layer name as key
reg_points = {}
for layer in viewer.layers:
    if layer.name.startswith('ring_'):
        reg_points[layer.name] = layer.data
        print(layer.name)
        
# iterate rings from inner to outer to construct numpy array of matching points
mapping = []
for i in range(n_rings):
    key_proj = f'ring_{i}_proj'
    key_gt = f'ring_{i}_gt'

    mapping_i = np.stack((reg_points[key_proj], reg_points[key_gt]), axis=1)
    # get rid of out of canvas points
    if i in excl_indices.keys():
        mapping_i = np.delete(mapping_i, excl_indices[i], axis=0)
    mapping.append(mapping_i)

# points-replicates-dim(101) - proj_vs_gt-dim(2) - YX-dim(2)
mapping = np.concatenate(mapping, axis=0)
mapping
```

```python
import numpy as np
from scipy.ndimage import generic_filter, gaussian_filter
from scipy.signal import convolve2d
from matplotlib import pyplot as plt

def make_cb(board_size):
    # Create a checkerboard pattern as the input image
    checkerboard = np.zeros(board_size, dtype=np.float32)
    squares = 2
    square_size = board_size[0] // squares

    for i in range(squares):
        for j in range(squares):
            if (i + j) % 2 == 0:
                checkerboard[i * square_size:(i + 1) * square_size, j * square_size:(j + 1) * square_size] = 1
    return np.stack((checkerboard,checkerboard,checkerboard), axis=2)

def make_coo_mesh(width=10, height=6):
    gridsize = 2
    # shape: (yx, Nrows, Ncols)
    grid = np.moveaxis(np.mgrid[0:width+1:gridsize,0:height+1:gridsize], 0,2).astype(float)
    print(grid.shape)
    n_rows, n_cols = grid.shape[1:]
    grid_flat = grid.reshape((-1,2))
    print(grid_flat)
    return grid,grid_flat

def show_cb(cb):
    plt.imshow(cb)
    plt.show()

def gkern(l=5, sig=1.):
    """
    creates gaussian kernel with side length `l` and a sigma of `sig`
    """
    ax = np.linspace(-(l - 1) / 2., (l - 1) / 2., l)
    gauss = np.exp(-0.5 * np.square(ax) / np.square(sig))
    kernel = np.outer(gauss, gauss)
    return kernel / np.sum(kernel)

def smooth_array_coos(arr, size, ignore_val=0.):
    kernel = gkern(size, )
    # print(kernel)
    def coordinate_smoothing(arr):
        arr = arr.reshape(size,size)
        if arr[15,15] == ignore_val:
            return ignore_val
        # print(arr)
        mask = arr != ignore_val
        ignored_propor = 1-kernel[mask].sum()
        convolved = (arr*kernel)[mask].sum()
        # print(ignored_propor)
        # print((arr*kernel))
        # print((arr*kernel)[mask])
        # print((arr*kernel)[mask].sum())
        # print()
        return convolved + ignored_propor*convolved
    print(arr.shape)
    new_arr = generic_filter(arr, coordinate_smoothing, size=(size,size), mode='constant', cval=-1)
    print(new_arr.shape)
    return new_arr
```

```python
# this cell generates a list of mesh coordiates in a 1080p image and transforms each 
# vertex using the `tform` object. This is the same process as the `warp()` example
# above, but here we get the exact projetion coordiates of the collection of vertices.
# The loop below is quite slow because each vertex is warped seperately as part of a
# (1080,1920) numpy array. `warp_coords()` didn't produce the expected output

# generate the mesh and convert to a list of yx vertex coordinates 
# smaller grids give better resolution, but runtime scales quadratically 
gridsize = 20
# shape: (yx, Nrows, Ncols)
grid = np.mgrid[0:1080+1:gridsize,0:1920+1:gridsize] 
n_rows, n_cols = grid.shape[1:]
# flatten - shape: (N, yx) 
grid_coos = grid.reshape((2, grid.size//2)).swapaxes(0,1) 

# create empty canvcas
canvas_base = np.zeros([1080, 1920], bool)
input_warped_canvas = np.zeros([1080, 1920], int)
output_warped_canvas = np.zeros([1080, 1920], int)

# itereate vertices
meshgrid = []
for i, grid_coo in enumerate(grid_coos):
    # draw an 8x8 blob of white on empty canvas copy
    canvas = canvas_base.copy()
    canvas[grid_coo[0]-4:grid_coo[0]+4, grid_coo[1]-4:grid_coo[1]+4] = True
    
    # warp, and calcuate two arrays (for x and y coo) capturing the location of the projected blob
    warped_coo = np.where(warp(canvas, tform))
    # get the center coo of the distorted, projected blob, if outside of canvas set to -1, -1
    warped_y, warped_x = np.median(np.stack(warped_coo), 1).astype(int)
    if not warped_coo[0].size:
        warped_y, warped_x = [-1,-1]
    
    # debugging: draw blobs and projections onto validation canvas
    x_cross = [x for x in range(warped_x-5,warped_x+5) if x<1920]
    y_cross = [y for y in range(warped_y-5,warped_y+5) if y<1080]
    output_warped_canvas[warped_y, x_cross] = i+1
    output_warped_canvas[y_cross, warped_x] = i+1
    input_warped_canvas[canvas] = i+1
    
    # save: input_vertex_xy, warped_vertex_xy
    meshpoint = np.array([grid_coo[-1::-1], [warped_x,warped_y]])
    meshgrid.append(meshpoint)
    print(grid_coo)
    
# n_rows*n_cols(5335)-dim - input_vtx-warped_vtx-dim - xy-dim
meshmapping = np.stack(meshgrid, axis=0)
```

```python
# in this cell the meshmapping is used to write an .obj file 
# integrating geometric vertices with uv texture mapping

def create_obj_file(vertices, texture_vertices, out_filename, flip_texture=False):
    # create all the vertex, vertex texture and vertex normals lists
    v, vt, vn, tr = [], [], [], []
    for i in range(0, vertices.shape[0]):
        # warped_vertex_i xy --> v (geometry/ 3D model)
        # normalize to 0-1, maintain aspect 16/9 ratio
        v.append((vertices[i] /np.array([1920,1920]) ))
        # inputmesh_vertex_i xy --> vt (texture uv map)
        # normalized to 0-1, change aspect ratio to 1/1
        # flip top and bottom by subtracting 1, then multiplying by -1
        vt.append((texture_vertices[i] /np.array([1920,1080])))
        if flip_texture:
            vt[-1] = (vt[-1] - np.array([0,1])) *np.array([1,-1])
        # vn (vertex normals) constant for all vertices
        vn.append(np.array([1, 0, 0]))

    # create polygons referencing elements above, skip last column for avoiding indexing error
    for i in range(0, meshmapping.shape[0]-n_cols):
        # the last vertex in a row should be skipped because no triangle can be drawn to the right
        if (i+1)%n_cols:
            topleft_i = i
            topright_i = i+1
            bottomleft_i = i+n_cols
            bottomright_i = i+1+n_cols

            # only add polygons that are based on valid in-canvas projections (not -1,-1)
            if not (np.stack([v[topright_i], v[bottomright_i], v[bottomleft_i], v[topleft_i]]).flatten() < 0).any():#         
#             if True:
                tr.append([topright_i, bottomright_i , bottomleft_i])
                tr.append([topleft_i, topright_i, bottomleft_i])

    # for file writing convert lists to strings 
    v_str = '\n'.join([f'v {v_[0]} {v_[1]} 0' for v_ in v])
    vt_str = '\n'.join([f'vt {vt_[0]} {vt_[1]} 0' for vt_ in vt])
    vn_str = '\n'.join(['vn 0 0 1' for _ in vn])
    triangles_str = '\n'.join([(f'f {tr_[0]+1}/{tr_[0]+1}/{tr_[0]+1}'
                                f' {tr_[1]+1}/{tr_[1]+1}/{tr_[1]+1}'
                                f' {tr_[2]+1}/{tr_[2]+1}/{tr_[2]+1}') for tr_ in tr])
    
    # write the .obj file
    with open(out_filename, 'w') as f:
        f.writelines(v_str + '\n')
        f.writelines(vt_str + '\n')
        f.writelines(vn_str + '\n')
        f.writelines(triangles_str + '\n')
```
